```
# ============================================================
# MAIN GITMAP WINDOW
# Creates the main PySide6 application window for GitMap.
# ============================================================

import sys

from PySide6.QtCore import Qt
from PySide6.QtGui import QKeySequence, QShortcut
from PySide6.QtWidgets import (
    QApplication,
    QLabel,
    QMainWindow,
    QPushButton,
    QSplitter,
    QTreeWidget,
    QTreeWidgetItem,
    QVBoxLayout,
    QWidget,
)


def main():
    """Launch the temporary GitMap PySide6 test window."""

    # --- Qt Application ---
    # Creates the application that manages the GUI.
    app = QApplication(sys.argv)

    # --- Application Identity ---
    # Gives Qt a consistent application and organization name.
    # These names can later be used for saved GUI settings.
    app.setApplicationName("GitMap")

    app.setOrganizationName("GitMap")

    # --- Application Font ---
    # Sets the default font size used throughout the GitMap GUI.
    app_font = app.font()
    app_font.setPointSize(15)
    app.setFont(app_font)

    # --- Main Window ---
    # Creates GitMap's temporary Qt window.
    window = QMainWindow()

    # --- Fullscreen Toggle ---
    # Switches GitMap between fullscreen and normal window mode.
    def toggle_fullscreen():
        if window.isFullScreen():
            window.showMaximized()
        else:
            window.showFullScreen()

    # --- Application Identity ---
    # Sets the title displayed by the operating system.
    window.setWindowTitle("GitMap")

    # --- Application Status Bar ---
    # Displays messages about what GitMap is currently doing.
    status_bar = window.statusBar()

    # --- Initial Status Message ---
    # Shows that GitMap is ready for the user.
    status_bar.showMessage("Ready")

    # --- Fullscreen Shortcut ---
    # Allows F11 to enter or leave fullscreen mode.
    window.keyPressEvent = lambda event: (
        toggle_fullscreen()
        if event.key() == Qt.Key.Key_F11
        else QMainWindow.keyPressEvent(window, event)
    )

    # --- Initial Window Size ---
    # Gives the test window a comfortable starting size.
    window.showFullScreen()

    # --- Main Workspace ---
    # Creates three horizontally arranged, resizable workspace areas.
    workspace = QSplitter(Qt.Orientation.Horizontal)

    # --- Splitter Handles ---
    # Makes the draggable dividers easier to find and grab.
    workspace.setHandleWidth(8)

    # --- Roadmap Navigation ---
    # Creates the left panel that will contain the roadmap opener
    # and the hierarchy of the currently opened roadmap.
    navigation_area = QWidget()

    # --- Navigation Minimum Width ---
    # Prevents the navigation panel from becoming unusably narrow.
    navigation_area.setMinimumWidth(200)

    # --- Navigation Layout ---
    # Arranges navigation controls vertically from top to bottom.
    navigation_layout = QVBoxLayout(navigation_area)

    # --- Open Roadmap Button ---
    # Opens a roadmap file and later displays the opened roadmap's name.
    roadmap_button = QPushButton("Open Roadmap")

    # --- Temporary Exit Button ---
    # Provides an easy way to close GitMap while fullscreen
    # window controls are hidden.
    exit_button = QPushButton("Exit GitMap")

    # --- Exit Button Action ---
    # Closes the GitMap application when clicked.
    exit_button.clicked.connect(window.close)

    # --- Exit Button Layout ---
    # Places the temporary exit button below Open Roadmap.
    navigation_layout.addWidget(exit_button)

    # --- Open Roadmap Shortcut ---
    # Gives Open Roadmap the standard Ctrl+O keyboard shortcut.
    open_roadmap_shortcut = QShortcut(
        QKeySequence.StandardKey.Open,
        window,
    )

    # --- Open Roadmap Shortcut Action ---
    # Makes Ctrl+O perform the same action as clicking Open Roadmap.
    open_roadmap_shortcut.activated.connect(roadmap_button.click)

    # --- Open Roadmap Button Layout ---
    # Places the roadmap opener at the top of the navigation panel.
    navigation_layout.addWidget(roadmap_button)

    # --- Roadmap Tree ---
    # Displays the hierarchy of the currently opened roadmap.
    roadmap_tree = QTreeWidget()

    # --- Roadmap Tree Header ---
    # Gives the hierarchy column a useful name.
    roadmap_tree.setHeaderLabel("Roadmap Contents")

    # --- Roadmap Tree Layout ---
    # Places the hierarchy underneath the Open Roadmap button
    # and allows it to use the remaining navigation space.
    navigation_layout.addWidget(roadmap_tree)

    # --- Temporary Milestone ---
    # Adds one temporary milestone to test the roadmap hierarchy.
    test_milestone = QTreeWidgetItem(
        roadmap_tree,
        ["0.1 Test Milestone"],
    )

    # --- Temporary Section ---
    # Adds a section underneath the temporary milestone.
    test_section = QTreeWidgetItem(
        test_milestone,
        ["0.1.1 Test Section"],
    )

    # --- Temporary Issue ---
    # Adds an issue underneath the temporary section.
    test_issue = QTreeWidgetItem(
        test_section,
        ["0.1.1.0.1 Test Issue"],
    )

    # --- Expand Test Hierarchy ---
    # Opens the temporary hierarchy so all levels are visible.
    roadmap_tree.expandAll()

    # --- Main Workspace ---
    # Creates three horizontally arranged, resizable workspace areas.
    workspace = QSplitter(Qt.Orientation.Horizontal)

    # --- Splitter Handles ---
    # Makes the draggable dividers easier to find and grab.
    workspace.setHandleWidth(8)

    # --- Item Editor ---
    # Center area will contain controls for editing the selected item.
    editor_area = QLabel("Item Editor")

    # Keeps enough room available for editing controls.
    editor_area.setMinimumWidth(400)

    # --- Live Preview ---
    # Right side will show the resulting roadmap content.
    preview_area = QLabel("Live Preview")

    # Keeps the preview readable when resizing the workspace.
    preview_area.setMinimumWidth(250)

    # --- Assemble Workspace ---
    # Adds each workspace area to the draggable splitter.
    workspace.addWidget(navigation_area)

    workspace.addWidget(editor_area)

    workspace.addWidget(preview_area)

    # --- Initial Workspace Sizes ---
    # Gives the editor the most room while keeping both side panels visible.
    workspace.setSizes(
        [
            250,
            700,
            350,
        ]
    )

    # --- Main Window Content ---
    # Makes the three-column workspace fill the GitMap window.

    window.setCentralWidget(workspace)
    # --- Application Event Loop ---
    # Keeps Qt running until the window is closed.
    sys.exit(app.exec())


# ============================================================
# MODULE ENTRY POINT
# Starts this test when the file is run directly.
# ============================================================

if __name__ == "__main__":
    # --- Launch GitMap ---
    # Starts the temporary PySide6 application.
    main()
```
